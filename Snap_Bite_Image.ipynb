# Snap-Bite – Image Ingredient Detection

This notebook is the image-only prototype for Snap-Bite.
It follows the Computer Vision / pretrained YOLO approach used in the course lab.

**Flow:** Image → Object Detection → Ingredient List


## 1. Install / Import Libraries


```python
!pip install ultralytics

from ultralytics import YOLO
from PIL import Image
import matplotlib.pyplot as plt
```

## 2. Load a Pretrained YOLO Model

This follows the pretrained YOLO approach from the course Computer Vision lab. For the first prototype, we use a pretrained model before building a custom food dataset.

```python
model = YOLO("yolov8l-worldv2.pt")
```

## 3. Define Snap-Bite Ingredient Classes

These are the ingredient categories we want the model to look for in the image.

```python
ingredients = [
    "milk", "egg", "cheese", "chicken", "beef", "fish",
    "tomato", "potato", "onion", "carrot", "cucumber",
    "bell pepper", "lettuce", "apple", "banana", "orange",
    "bread", "butter", "yogurt"
]

model.set_classes(ingredients)
```

## 4. Load a Refrigerator Image


```python
image_path = "test_fridge.jpg"
image = Image.open(image_path)
display(image)
```

## 5. Detect Ingredients


```python
results = model(image_path)
annotated_image = results[0].plot()

plt.figure(figsize=(12, 8))
plt.imshow(annotated_image[:, :, ::-1])
plt.axis("off")
plt.show()
```

## 6. Extract the Detected Ingredient Names


```python
detected_ingredients = []

for result in results:
    for cls_id in result.boxes.cls.tolist():
        name = result.names[int(cls_id)]
        if name not in detected_ingredients:
            detected_ingredients.append(name)

print("Detected ingredients:")
for ingredient in detected_ingredients:
    print("-", ingredient)
```
